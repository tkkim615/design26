```python
df = pd.read_csv('Color.csv', header=0, index_col=None)
df = df.replace(" ", np.nan)
df = df.fillna(0.0)
df
```

```python
xy_data = np.array(df, dtype=np.float32)
xy_data
```

```python
x_data = xy_data[:, :-1]
y_data = xy_data[:, [-1]]

print(x_data.shape, y_data.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_data, y_data, train_size=0.8, shuffle=True)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=10, random_state=None)
rf_model.fit(x_train, y_train.ravel())
```

```python
predict_train_rf = rf_model.predict(x_train)
predict_test_rf = rf_model.predict(x_test)

plt.figure(figsize=(6,5))

plt.scatter(y_train, predict_train_rf, color='blue', label='Train Set Predictions', alpha=0.6)
plt.scatter(y_test, predict_test_rf, color='red', label='Test Set Predictions', alpha=0.6)

plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.legend()
plt.grid(True)
plt.show()
```

```python
feature_importances = rf_model.feature_importances_
feature_names = df.columns[:-1]

importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': feature_importances})
importance_df = importance_df.sort_values(by='Importance', ascending=False)

plt.figure(figsize=(6,4))
plt.bar(importance_df['Feature'], importance_df['Importance'])
plt.xlabel('Features')
plt.ylabel('Importance')
plt.title('Feature Importances for Random Forest Regressor')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(15,5))

plot_tree(rf_model.estimators_[0],
          feature_names=df.columns[:-1].tolist(),
          filled=True,
          rounded=True,
          max_depth=3,          # Limit depth
          fontsize=8)

plt.show()
```
