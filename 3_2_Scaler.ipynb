```python
import numpy as np
```

```python
rand_no = np.random.normal(0,100,10)       # 평균 0, 표준편차 100의 정규분포를 가지는 10개의 숫자 생성
print(rand_no)
print(rand_no.shape)
```

```python
rand_no = rand_no.reshape(-1,1)
print(rand_no)
print(rand_no.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

MMscaler = MinMaxScaler(feature_range=(-1,1))                                              # MinMaxScaler()도 가능, 결과 같음
rand_no_MinMax = MMscaler.fit_transform(rand_no)
print(rand_no_MinMax)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(5, 4))
plt.scatter(rand_no.flatten(), rand_no_MinMax.flatten())
plt.grid()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

STscaler = StandardScaler()                                      # 평균 0, 표준편차 1이 되도록 scaling
rand_no_STscaler = STscaler.fit_transform(rand_no)
print(rand_no_STscaler)

mean_st_scaler = np.mean(rand_no_STscaler)
std_st_scaler = np.std(rand_no_STscaler)
print(f"rand_no_STscaler의 평균: {mean_st_scaler:.4f}")
print(f"rand_no_STscaler의 표준 편차: {std_st_scaler:.4f}")
```
