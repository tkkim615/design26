```python
data = [[5,1,10,8,15],
        [2,1,8,9,21],
        [3,2,1,10,29],
        [5,4,9,6,38],
        [4,4,5,7,52],
        [5,4,6,4,63],
        [6,5,2,5,70],
        [9,7,4,1,78],
        [7,7,3,2,91],
        [9,8,3,3,95]]
data
```

```python
import numpy as np
xy_data = np.array(data, dtype=np.float32)
xy_data
```

```python
x_data = xy_data[:, :-1]            # x_data = xy_data[0:10, 0:4]
y_data = xy_data[:, [-1]]           # y_data = xy_data[0:10, -1]
print(x_data)
print(y_data)
print(x_data.shape)
print(y_data.shape)
```

```python
import tensorflow as tf
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Input
from keras.optimizers import Adam

model = Sequential()
model.add(Input(shape=(4,)))
model.add(Dense(1, activation = 'linear'))
model.compile(optimizer=Adam(learning_rate=5.0), loss = 'mse')
model.summary()
```

```python
hist = model.fit(x_data, y_data, epochs=100)
```

```python
import matplotlib.pyplot as plt

plt.plot(hist.history['loss'])
plt.show()
```

```python
y_predict = model.predict(x_data)
y_predict
```

```python
plt.plot(y_data, y_predict, 'o')
plt.xlabel('y_data')
plt.ylabel('y_predict')
plt.show()
```

```python
x_test = np.array([[7,6,2,9]])
predict_score = model.predict(x_test)
print("예상점수:", predict_score)
```
