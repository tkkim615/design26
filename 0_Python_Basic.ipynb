```python
def calculate_diagonal(width, height):
    diagonal_length = (width**2 + height**2)**0.5     # 제곱근 연산자 **
    return diagonal_length

width_val = 3
height_val = 4
diagonal = calculate_diagonal(width_val, height_val)

print("대각선 길이:", diagonal)
```
